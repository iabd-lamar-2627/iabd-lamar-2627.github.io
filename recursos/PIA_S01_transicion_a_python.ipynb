## 2 · Colecciones

Cuatro estructuras que vas a usar constantemente. La diferencia entre ellas no es cosmética: determina qué operaciones son rápidas y cuáles no.

| Tipo | Se escribe | Ordenada | Modificable | Sirve para |
|---|---|---|---|---|
| `list` | `[1, 2, 3]` | sí | sí | secuencias de cualquier cosa |
| `tuple` | `(1, 2, 3)` | sí | **no** | registros fijos: coordenadas, dimensiones |
| `dict` | `{"a": 1}` | sí (por inserción) | sí | pares clave-valor, búsqueda instantánea |
| `set` | `{1, 2, 3}` | no | sí | pertenencia y eliminación de duplicados |

````python
detecciones = ["as", "rey", "as", "sota", "rey", "as"]

print("Lista       ", detecciones)
print("Únicas      ", set(detecciones))
print("Nº únicas   ", len(set(detecciones)))
print("¿hay sota?  ", "sota" in detecciones)

# Un dict de conteo, a mano
conteo = {}
for carta in detecciones:
    conteo[carta] = conteo.get(carta, 0) + 1
print("Conteo      ", conteo)

# Tupla: una caja fija (ancho, alto) que nadie debe modificar
resolucion = (1280, 720)
ancho, alto = resolucion          # desempaquetado
print(f"Resolución   {ancho}x{alto}")
````

## 3 · Recorrer cosas

En Python **no se recorre por índice** salvo que el índice haga falta de verdad. Se recorre el objeto directamente. Si necesitas la posición, `enumerate`. Si necesitas recorrer dos secuencias a la vez, `zip`.

````python
cartas = ["as", "rey", "sota"]
valores = [1, 12, 10]

print("— for directo")
for c in cartas:
    print(" ", c)

print("\n— con enumerate (posición + valor)")
for i, c in enumerate(cartas, start=1):
    print(f"  {i}. {c}")

print("\n— con zip (dos secuencias a la vez)")
for c, v in zip(cartas, valores):
    print(f"  {c:<5} vale {v}")
````

## 4 · Comprensiones

Esta es la construcción que más separa el código Python del código de otros lenguajes, y la que más vas a ver en el curso. Una comprensión construye una colección a partir de otra en una sola línea.

```python
[ expresión  for elemento in iterable  if condición ]
```

Se lee de izquierda a derecha: *«dame la expresión, para cada elemento del iterable, si se cumple la condición»*.

````python
nums = [1, 2, 3, 4, 5, 6, 7, 8]

print("Clásico:")
pares = []
for n in nums:
    if n % 2 == 0:
        pares.append(n * n)
print(" ", pares)

print("Comprensión:")
pares = [n * n for n in nums if n % 2 == 0]
print(" ", pares)

# También hay comprensiones de diccionario y de conjunto
cartas = ["as", "rey", "sota"]
print("Dict: ", {c: len(c) for c in cartas})
print("Set:  ", {len(c) for c in cartas})
````

## 5 · Funciones

Lo esencial: argumentos por defecto, argumentos con nombre, y **anotaciones de tipo**. Las anotaciones no obligan a nada en ejecución, pero documentan la función y las vas a encontrar en todo el código de bibliotecas.

````python
def clasificar(confianza: float, umbral: float = 0.5) -> str:
    """Devuelve 'aceptada' o 'descartada' según el umbral dado."""
    return "aceptada" if confianza >= umbral else "descartada"


print(clasificar(0.77))                  # usa el umbral por defecto
print(clasificar(0.77, 0.8))             # por posición
print(clasificar(0.77, umbral=0.8))      # por nombre, más legible
print(clasificar.__doc__)
````

````python
# Una función puede devolver varios valores (en realidad devuelve una tupla)
def resumen(lecturas: list[float]) -> tuple[int, float, float]:
    return len(lecturas), min(lecturas), max(lecturas)

n, menor, mayor = resumen([0.91, 0.34, 0.77])
print(f"{n} lecturas, entre {menor} y {mayor}")
````

## 6 · Clases, lo justo

En este curso las clases se usan poco: casi todo son funciones y estructuras de datos. Pero necesitas reconocerlas. Lo mínimo es `__init__` (el constructor) y `__repr__` (cómo se imprime el objeto).

Para clases que solo guardan datos, Python tiene un atajo: `@dataclass`, que escribe el `__init__` y el `__repr__` por ti.

````python
from dataclasses import dataclass

@dataclass
class Deteccion:
    etiqueta: str
    confianza: float
    caja: tuple[int, int, int, int]

    def es_fiable(self, umbral: float = 0.5) -> bool:
        return self.confianza >= umbral


d = Deteccion("as", 0.91, (100, 120, 60, 90))
print(d)                     # __repr__ generado automáticamente
print(d.etiqueta, d.confianza)
print("¿fiable?", d.es_fiable(), "· con umbral 0.95:", d.es_fiable(0.95))
````

## 7 · Ficheros y rutas

Dos reglas que se aplican en todo el curso:

1. **Rutas con `pathlib`, no con cadenas.** `Path("datos") / "cartas.json"` funciona en Windows, Linux y macOS. `"datos/cartas.json"` te va a dar problemas.
2. **Ficheros siempre con `with`.** El fichero se cierra solo aunque salte una excepción a mitad.

````python
import json
from pathlib import Path

destino = Path("salida") / "cartas.json"
destino.parent.mkdir(parents=True, exist_ok=True)   # crea la carpeta si no existe

datos = {"cartas": ["as", "rey", "sota"], "valores": [1, 12, 10]}

with open(destino, "w", encoding="utf-8") as f:
    json.dump(datos, f, ensure_ascii=False, indent=2)

print("Escrito en:", destino.resolve())
print("Tamaño:", destino.stat().st_size, "bytes")
print("\nContenido:")
print(destino.read_text(encoding="utf-8"))
````

> **`encoding="utf-8"` no es opcional.** En Windows, Python usa por defecto la codificación del sistema y los acentos y las eñes se rompen. Ponlo siempre, en lectura y en escritura.
